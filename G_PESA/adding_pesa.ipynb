# Adding PESA — imaging patterns vs. behavioral patterns

**Experiment 2.** No subject-level overlap between cohorts: imaging = 25 unilateral OTC resection
patients + 38 controls; PESA = 20 hemispherectomy patients + 8 controls. Comparison is at the level
of *pattern*, never subject.

Two patterns are tested on both sides:

1. **bi/uni** — bilateral categories (object, house) vs lateralized (face, word). One score per subject.
2. **gradient** — a priori rank order `object = house > face > word`, fixed from control LI. Never
   re-derived from observed effect sizes.

Imaging conventions carried over: OTC -> last session; controls -> first session; exclude `sub-017`
and (`sub-108`, ses-02); split by intact hemisphere, never pooled. `hemi='l'` = LH-intact patients
vs controls' left.

`liu_distinctiveness` is a **similarity** metric: higher = *less* distinct.

## 0. Verified results — carry forward, do not re-derive

Established and confirmed. Anything below is a **result**, not a lead. Do not recompute to "check."

### The finding

| Measure | Result |
|---|---|
| **Behavior (PESA)** | Significant bilateral vs. lateralized difference, patients vs. controls. Trial-level GLMM, `lat_load = |LI|`, pre-registered 1-df term. See `pesa_now.ipynb` §7b. |
| **Distinctiveness — word** | RH-intact patients vs. matched controls: d = +0.914, q = .031. rVWFA less distinct. |
| **Distinctiveness — face** | Suggestive, LH > RH: LH-intact d = −0.761 (q = .083), RH-intact d = +0.605 (q = .099). Opposite signs. |
| **Convergence** | Distinctiveness and behavior agree: the *lateralized* categories (face, word) are the affected ones. |
| **Sum-selectivity** | Does NOT predict behavior. Its one significant per-category effect is object_LOC (bilateral), and its bi/uni gap runs opposite in sign to behavior's. |

### Established data facts

- **Cohorts do not overlap.** Imaging = 25 OTC resection patients (13 LH-intact / 12 RH-intact) + 38 controls. PESA = 20 hemispherectomy patients + 8 controls. **Zero shared subjects** — no subject-level brain-behavior link is possible. Group-level pattern comparison only.
- **PESA behavioral controls are incomplete** (n = 8, collection ongoing). Treat control-referenced behavioral numbers as provisional.
- **PESA test structure** (`pesa_now.ipynb` §7b, cell 40): `correct ~ group + stimCat + <term> + ageScaled + blockPos_c + gender + (1|subject_id)`, binomial. Two pre-registered 1-df terms: `vuln = resection_sign × LI` (signed) and `lat_load = |LI|` for patients (unsigned). `LI = {words:+1, faces:−1, cars/houses/hands:0}`; `RESECT_SIGN = {Control:0, Patient_Left:−1, Patient_Right:+1}`. Hypothesis β < 0 for both.
- **`lat_acc` = mean(faces, words); `bil_acc` = mean(houses, cars, hands); `gap_acc` = bil − lat.** Verified against raw per-category columns. `hands` is in the bilateral bin and has no imaging counterpart, so the behavioral bilateral bin is broader than the imaging one.
- **Signed-log transform replaces log10** for sum-selectivity: `sign(x) · log10(1 + |x|)`. Reason: ComBat pushes near-zero values negative (19 non-positive rows harmonized vs. 9 raw; up to 10 of 38 controls for word_VWFA), and `log10` silently dropped them. Signed log retains the full 38/13/12 in every cell. Headline unchanged: object_LOC LH-intact d = −1.28 either way.
- **`sum_selec_norm` cannot be negative at source** — `supra = mask & (z > 2.326)`, so only positive voxels are summed. All negatives are a ComBat artifact, not face-dominant parcels.
- **Distinctiveness is a similarity metric.** Higher = LESS distinct. diff = pt − ctrl. Positive = patients less distinct.
- **`control_li.py` and sum-selectivity use different inputs but agree.** LI runs on `n_selective` voxel counts from `group_results/roi_betas/selective_voxel_counts.csv` (thresholds 1.96/2.33, excludes sub-027/sub-084, MIN_TOTAL=10); sum-sel runs on summed z from `univariate_v1.csv` (threshold 2.326, excludes sub-017 and sub-108 ses-2). Control cohorts share 36 of 38. Computing |LI| from `sum_selec_norm` reproduces the same ordering (word .326 ≈ face .321 > house .207 > object .119), so the two measures are consistent. **Do not re-litigate this.**
- **The predicted rank order is a priori from control LI**: object = house > face > word. Never re-derive it from observed effect sizes.
- **Age is confounded with group.** Range 7.5–38.5 y, median 14.83. Patients skew older (17/8), controls younger (14/24). Ages are well matched *within* bin (younger 12.0 vs 11.4; older 19.0 vs 20.5). Use age as a covariate, not a median split. No scanner column in the harmonized files.

### Verified negatives — do not re-run

- **Pooled / merged control reference: abandoned.** Averaging both control hemispheres compares patients to a mean describing neither. Control L-vs-R differs significantly in 3/4 sum-sel ROIs, 2/4 distinctiveness ROIs, 3/4 geometry ROIs. Empirically, object_LOC LH-intact goes from p = .0002 (matched) to p = .040 (pooled). **Matched-hemisphere is primary. Not revisiting.**
- **Category-level correlation of 4 group means cannot reach significance** — 4! = 24 orderings, minimum two-tailed p = .083. Any test built on correlating 4 category means is uninformative by construction.
- **Patient subgroup LMMs (LH-intact vs RH-intact), OLS with subject-clustered SEs**: sum-selectivity χ²(3) = 6.080, p = .108; distinctiveness χ²(3) = 6.615, p = .085. Both null → no per-category follow-ups. Report as uninformative, not equivalence. MixedLM does not converge here (subject random-intercept variance ≈ 0.008 vs residual ≈ 0.22, unidentified with one observation per subject per category) — use OLS + cluster.
- **Crossed comparison** (patients vs opposite-hemisphere controls), BH-FDR within measure across 8 cells:
  - Sum-sel: object_LOC significant in both subgroups (LH d = −0.96, RH d = −1.06, both q = .011); face_FFA RH-intact vs left controls d = +1.11, q = .011 — but controls already differ L-vs-R on face_FFA, so this cell cannot separate patient effect from normal asymmetry.
  - Distinctiveness: nothing survives. word_VWFA RH-intact vs left controls d = +0.076, p = .517 — the patient value (0.731) sits near the control *left* mean (0.655), above control right (0.373). Matched and crossed shifts cancel; they do not add.
- **Ordering consistency** (per-subject ρ vs predicted order): patients trend lower but not significantly — LH pt +0.585 vs ctrl +0.800 (p = .120); RH pt +0.567 vs ctrl +0.726 (p = .138). No age relationship in any group (all p > .10).
- **Age median split on ordering is not reportable.** Younger patient cells are n = 4. The two apparent order switches each trace to one or two subjects with a near-zero parcel value (sub-008 house = −0.36 and sub-101 in pt_l older; sub-010 face = −1.58 in pt_r younger). Not developmental.
- **Absolute-value ordering is identical in all four groups** (object > house > face > word, signed log), so absolute sum-selectivity cannot explain a patient–control behavioral difference. Friedman omnibus significant in all four groups, but only the object > house step survives pairwise in patients; the face vs word step is the least reliable in 3 of 4 groups.
- **`pt_vs_0` tests on imaging bi/uni gaps are meaningless.** Zero implies object/house equal face/word in magnitude, which is false by parcel construction — controls are significant too (p < .0001). Only patient-vs-control is interpretable for imaging. Behavior is different: accuracy is on a common 0–1 scale, so its vs-zero test is valid.
- **Subject-centered sensitivity check** (each subject's values minus their own across-parcel mean): object_LOC LH-intact holds (d = −1.28 → −0.97, q = .026); no new effects; word stays null. Between-subject magnitude variation is not masking a word effect.
- **`lat_load` / `vuln` on imaging** (OLS + clustered SE, `y ~ C(group) + C(category) + term + age_z`): only sum-selectivity `lat_load` with ordinal LI reaches significance, β = **+0.416**, χ²(1) = 4.57, p = .033 — **positive, i.e. opposite to the pre-registered β < 0**. Likely re-expressing the object_LOC drop rather than a lateralized gain. `vuln` null everywhere (p = .31–.45). Distinctiveness null on both terms, both codings.

### Control reference values

Control |LI| (`control_li.py`, z > 1.96, n = 38): word 0.357 (signed +0.337, median +0.360) > face 0.234 (signed −0.191, median −0.203) > house 0.115 (signed −0.097) > object 0.114 (signed +0.098).

Control face − word sum-selectivity gap (signed log): face exceeds word in **both** hemispheres, but the gap is significantly larger on the right — L +0.181 (vs 0, p = .194) vs R +0.719 (vs 0, p = .0014); paired L vs R p = .017. Neither patient group's gap differs from zero (LH p = .542, RH p = .151). So face is always stronger than word; the *degree* is hemisphere-dependent.

Distinctiveness means (signed, harmonized): ctrl L face 0.680 / house 0.442 / object 1.060 / word 0.655; ctrl R face 0.536 / house 0.480 / object 1.219 / word 0.373; pt LH-intact 0.450 / 0.063 / 1.056 / 0.553; pt RH-intact 0.726 / 0.916 / 1.277 / 0.731. **house_PPA control SDs are ~0.71–0.77, roughly double every other parcel** — this is the variance problem, and house_PPA effect sizes are comparable to word_VWFA's (d ≈ 0.57–0.60 vs 0.91) while failing FDR.

## 1. Setup

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import spearmanr, wilcoxon, mannwhitneyu
import matplotlib.pyplot as plt

sys.path.insert(0, "/user_data/csimmon2/git_repos/sym_pt")
from sym_pt_params import processed_dir

np.random.seed(0)
D_LIU       = Path("/user_data/csimmon2/git_repos/sym_pt/D_liu")
UNIVAR_HARM = D_LIU / "univariate_v1_harmonized.csv"
RSA_HARM    = D_LIU / "rsa_v1_harmonized.csv"
PESA_CSV    = Path("/user_data/csimmon2/git_repos/sym_pt/G_PESA/PESA_master_plus_results.csv")
FIG_DIR     = Path("/user_data/csimmon2/git_repos/sym_pt/C_results/figures")
FIG_DIR.mkdir(parents=True, exist_ok=True)

ROIS        = ["face_FFA", "house_PPA", "object_LOC", "word_VWFA"]
CATS4       = ["face", "house", "object", "word"]
ROI_FOR     = dict(zip(CATS4, ROIS))
HEMIS       = {"l": "LH-intact", "r": "RH-intact"}
EXCLUDE     = ["sub-017"]
EXCLUDE_SES = [("sub-108", 2)]
N_PERM      = 10_000
MIN_CATS    = 3          # categories a subject needs for a rho / gap score

# bi/uni bins
BILATERAL   = ["object", "house"]
LATERALIZED = ["face", "word"]

# A priori gradient, fixed from control LI: object = house > face > word.
# Higher rank = predicted GREATER loss, so profiles below are built as
# (control mean - patient), i.e. deficit. Never re-derive this order.
PRED_RANK   = {"object": 3.5, "house": 3.5, "face": 2.0, "word": 1.0}
```

## 2. Load imaging — `uni`, `rsa`
Same exclusion + session rule for both files. `sum_selec_norm` from the univariate file,
`liu_distinctiveness` from the RSA file (collapsed over `pair`).

```python
def load_harmonized(path):
    d = pd.read_csv(path)
    d = d[~d["subject_id"].isin(EXCLUDE)].copy()
    d["ses_num"] = pd.to_numeric(d["session"], errors="coerce").astype(int)
    for bad_sub, bad_ses in EXCLUDE_SES:
        d = d[~((d["subject_id"] == bad_sub) & (d["ses_num"] == bad_ses))]
    if "pair" in d.columns:                       # rsa is long by pair; collapse
        d = d.drop(columns=["pair", "fisher_r"]).drop_duplicates()
    pt = d[d["group"] == "OTC"]
    ls = pt.groupby("subject_id")["ses_num"].max().reset_index().rename(columns={"ses_num": "keep"})
    pt = pt.merge(ls, on="subject_id"); pt = pt[pt.ses_num == pt.keep].drop(columns="keep")
    ct = d[d["status"] == "control"]
    fs = ct.groupby("subject_id")["ses_num"].min().reset_index().rename(columns={"ses_num": "keep"})
    ct = ct.merge(fs, on="subject_id"); ct = ct[ct.ses_num == ct.keep].drop(columns="keep")
    out = pd.concat([pt, ct], ignore_index=True)
    out["is_pt"] = (out["group"] == "OTC").astype(int)
    return out[out["category"].isin(ROIS)].copy()

uni = load_harmonized(UNIVAR_HARM)
uni["value"] = np.log10(uni["sum_selec_norm"].where(uni["sum_selec_norm"] > 0))

rsa = load_harmonized(RSA_HARM)
rsa["value"] = rsa["liu_distinctiveness"]

for nm, d in [("sum-sel", uni), ("distinctiveness", rsa)]:
    n = d.drop_duplicates("subject_id").groupby(["is_pt", "hemi"]).size()
    print(f"{nm}: {dict(n)}")
```

## 3. Helpers

```python
def cohen_d(x, y):
    nx, ny = len(x), len(y)
    sp = np.sqrt(((nx-1)*np.var(x, ddof=1) + (ny-1)*np.var(y, ddof=1)) / (nx+ny-2))
    return np.nan if sp == 0 else (np.mean(x) - np.mean(y)) / sp

def perm_test(x, y, n=N_PERM, rng=None):
    rng = rng or np.random.default_rng(0)
    obs = np.mean(x) - np.mean(y)
    pool = np.concatenate([x, y]); nx = len(x)
    null = np.empty(n)
    for i in range(n):
        rng.shuffle(pool); null[i] = np.mean(pool[:nx]) - np.mean(pool[nx:])
    return obs, (np.sum(np.abs(null) >= abs(obs)) + 1) / (n + 1)

def bh_fdr(p):
    p = np.asarray(p, float); m = len(p); order = np.argsort(p)
    adj = np.empty(m); run = 1.0
    for k in range(m-1, -1, -1):
        i = order[k]; run = min(run, p[i] * m / (k+1)); adj[i] = run
    return adj

def stars(p):
    if np.isnan(p): return ""
    return "***" if p < .001 else "**" if p < .01 else "*" if p < .05 else ""

def wide_by_category(d):
    """One row per subject x hemi, one column per category. Reference for
    both the bi/uni score and the gradient."""
    w = (d.pivot_table(index=["subject_id", "hemi", "is_pt"],
                       columns="category", values="value", aggfunc="first")
           .rename(columns={v: k for k, v in ROI_FOR.items()})
           .reset_index())
    return w
```

## 4. bi/uni — imaging (roadmap 2c, 3c)

Per subject: `gap = mean(bilateral) - mean(lateralized)`, matching PESA's `gap_acc`. Patients are
then compared to their **same-hemisphere** controls (10k perm), and the gap is tested against zero
within each group. Requires all four categories present, so subjects with a missing parcel drop.

```python
def biuni_scores(d):
    """gap = mean(bilateral) - mean(lateralized), per subject x hemi.
    Tolerates a missing parcel: needs >=1 category in each bin."""
    w = wide_by_category(d)
    have = [c for c in CATS4 if c in w.columns]
    bil = [c for c in BILATERAL   if c in have]
    lat = [c for c in LATERALIZED if c in have]
    w["n_cats"] = w[have].notna().sum(axis=1)
    w["bil"] = w[bil].mean(axis=1, skipna=True)
    w["lat"] = w[lat].mean(axis=1, skipna=True)
    w = w[w[bil].notna().any(axis=1) & w[lat].notna().any(axis=1)]
    w["gap"] = w["bil"] - w["lat"]
    return w.dropna(subset=["gap"])

rng = np.random.default_rng(0)
rows = []
for nm, d in [("sum_selectivity", uni), ("distinctiveness", rsa)]:
    w = biuni_scores(d)
    for h, hlab in HEMIS.items():
        x = w[(w.hemi == h) & (w.is_pt == 1)]["gap"].values
        y = w[(w.hemi == h) & (w.is_pt == 0)]["gap"].values
        if len(x) < 2 or len(y) < 2: continue
        diff, p = perm_test(x, y, rng=rng)
        rows.append(dict(measure=nm, group=hlab, n_pt=len(x), n_ct=len(y),
                         gap_pt=x.mean(), gap_ct=y.mean(), diff=diff,
                         d=cohen_d(x, y), p_vs_ctrl=p,
                         p_pt_vs_0=wilcoxon(x)[1], p_ct_vs_0=wilcoxon(y)[1]))
biuni_img = pd.DataFrame(rows)
biuni_img["q_vs_ctrl"] = bh_fdr(biuni_img["p_vs_ctrl"])
biuni_img.round(4)
```

## 5. Gradient — imaging (roadmap 2d, 3d, reproduced here for parity)

Per patient: profile of (value - same-hemisphere control mean) across the four categories,
correlated with `PRED_RANK`. **The predicted order is fixed a priori** — it is never re-derived
from the observed effect sizes. One-sample Wilcoxon of the per-patient rho against zero.

```python
def gradient_rhos(d, min_cats=MIN_CATS):
    """Per patient: Spearman rho between the DEFICIT profile
    (same-hemisphere control mean - patient) and PRED_RANK.
    Uses whichever categories the subject has (>= min_cats)."""
    w = wide_by_category(d)
    have = [c for c in CATS4 if c in w.columns]
    out = {}
    for h in HEMIS:
        ctrl = w[(w.hemi == h) & (w.is_pt == 0)][have].mean()
        pt   = w[(w.hemi == h) & (w.is_pt == 1)]
        rhos = []
        for _, r in pt.iterrows():
            cats = [c for c in have if pd.notna(r[c])]
            if len(cats) < min_cats: continue
            deficit = [ctrl[c] - r[c] for c in cats]     # positive = loss
            pred    = [PRED_RANK[c] for c in cats]
            if len(set(pred)) < 2: continue
            rho = spearmanr(deficit, pred)[0]
            if not np.isnan(rho): rhos.append(rho)
        out[h] = np.array(rhos)
    return out

rows = []
for nm, d in [("sum_selectivity", uni), ("distinctiveness", rsa)]:
    for h, r in gradient_rhos(d).items():
        rows.append(dict(measure=nm, group=HEMIS[h], n=len(r),
                         mean_rho=r.mean(), median_rho=np.median(r),
                         p=wilcoxon(r)[1] if len(r) > 5 else np.nan))
grad_img = pd.DataFrame(rows)
grad_img.round(4)
```

## 7. bi/uni — behavior

```python
rows = []
ct = pesa[pesa.is_pt == 0]["gap_acc"].dropna().values
for lab, sel in [("All patients", pesa.is_pt == 1),
                 ("LH-intact",    (pesa.is_pt == 1) & (pesa.hemi == "l")),
                 ("RH-intact",    (pesa.is_pt == 1) & (pesa.hemi == "r"))]:
    x = pesa[sel]["gap_acc"].dropna().values
    u, p_mw = mannwhitneyu(x, ct, alternative="two-sided")
    rows.append(dict(group=lab, n=len(x), gap=x.mean(), median=np.median(x), sd=x.std(),
                     p_vs_0=wilcoxon(x)[1], U=u, p_vs_ctrl=p_mw))
rows.append(dict(group="Control", n=len(ct), gap=ct.mean(), median=np.median(ct),
                 sd=ct.std(), p_vs_0=wilcoxon(ct)[1], U=np.nan, p_vs_ctrl=np.nan))
biuni_beh = pd.DataFrame(rows)
biuni_beh.round(4)
```

## 8. Gradient — behavior

Same construction as §5: per-patient accuracy profile minus the control mean, correlated with the
same a priori `PRED_RANK`. Category mapping `cars -> object` (hands has no imaging analog and is
excluded here, though it *is* inside `bil_acc` above — an asymmetry worth stating in Methods).

```python
BEH_MAP  = {"cars": "object", "houses": "house", "faces": "face", "words": "word"}
acc_cols = {b: f"{b}_acc" for b in BEH_MAP}

# deficit = control mean - patient accuracy (positive = worse), matching
# the imaging construction in section 5.
ctrl_mean = {c: pesa.loc[pesa.is_pt == 0, acc_cols[b]].mean() for b, c in BEH_MAP.items()}
pred      = [PRED_RANK[c] for c in BEH_MAP.values()]

rows = []
for lab, sel in [("LH-intact", pesa.hemi == "l"), ("RH-intact", pesa.hemi == "r")]:
    sub  = pesa[(pesa.is_pt == 1) & sel]
    rhos = []
    for _, r in sub.iterrows():
        deficit = [ctrl_mean[c] - r[acc_cols[b]] for b, c in BEH_MAP.items()]
        if np.any(pd.isna(deficit)): continue
        rhos.append(spearmanr(deficit, pred)[0])
    rhos = np.array(rhos)
    rows.append(dict(group=lab, n=len(rhos), mean_rho=rhos.mean(),
                     median_rho=np.median(rhos), p=wilcoxon(rhos)[1]))
grad_beh = pd.DataFrame(rows)
grad_beh.round(4)
```

## 9. Convergence table

The deliverable: does each imaging measure reproduce the behavioral pattern? Pattern-level only —
no subject-level correspondence is claimed or testable.

```python
def verdict(p, thr=.05):
    return "n/a" if (p is None or np.isnan(p)) else ("YES" if p < thr else "no")

# Two test types, reported separately so imaging and behavior are never
# compared across different tests:
#   vs_ctrl : patients vs their control group  (perm / Mann-Whitney)
#   vs_0    : is the bi/uni gap non-zero within group (Wilcoxon)
rows = []
for nm in ["sum_selectivity", "distinctiveness"]:
    b = biuni_img[biuni_img.measure == nm]
    g = grad_img[grad_img.measure == nm]
    rows.append(dict(
        measure        = nm,
        biuni_vs_ctrl  = ", ".join(f"{r.group}:{verdict(r.p_vs_ctrl)}" for r in b.itertuples()),
        biuni_pt_vs_0  = ", ".join(f"{r.group}:{verdict(r.p_pt_vs_0)}" for r in b.itertuples()),
        gradient       = ", ".join(f"{r.group}:{verdict(r.p)}"         for r in g.itertuples()),
    ))

_all = biuni_beh[biuni_beh.group == "All patients"].iloc[0]
_sub = biuni_beh[biuni_beh.group.isin(["LH-intact", "RH-intact"])]
rows.append(dict(
    measure        = "BEHAVIOR (PESA)",
    biuni_vs_ctrl  = ", ".join(f"{r.group}:{verdict(r.p_vs_ctrl)}" for r in _sub.itertuples()),
    biuni_pt_vs_0  = f"All:{verdict(_all.p_vs_0)}, " + ", ".join(
                        f"{r.group}:{verdict(r.p_vs_0)}" for r in _sub.itertuples()),
    gradient       = ", ".join(f"{r.group}:{verdict(r.p)}" for r in grad_beh.itertuples()),
))
convergence = pd.DataFrame(rows)
convergence
```
